## matplotlib图像结构介绍

Figure：画布，也就是整个图片窗口。
Axes：真正绘图的区域。注意这里虽然叫 Axes，但它不是“坐标轴”，而是一个完整的子图区域。如果一张图有多个子图，那么就有4个Axes
Axis：坐标轴对象，比如 x 轴和 y 轴。
Line2D：折线图中的线。
Text：标题（title）、坐标轴标签（xlabel）、刻度文字（tick label）文字注释。
Legend：图例。
Patch：矩形、柱子、圆形等图形对象。

## 基础绘图功能

```python
#绘制北京60分钟（x轴时间）以内的温度数据,温度区间15——18（y轴温度）
```

```python
#准备数据
x_time=range(0,60,1)
print(type(x_time))   #结果为一个可迭代对象range
y_tem=np.random.randint(0,10,60)
y_tem=np.random.uniform(15,18,60)
y_tem
```

```python
#绘制初始图像
#创建画布
plt.figure(figsize=(10,5),dpi=100)
#绘图
plt.plot(x_time,y_tem,color='red')
#显示图像
plt.show()
```

```python
#添加刻度，刻度标签、网格
#创建画布
plt.figure(figsize=(10,5),dpi=100)
#绘图
plt.plot(x_time,y_tem,color='red')
#添加刻度
x_tick_label=[f'11点{i}分' for i in range(0,60,5)]
y_tick=range(10,30,5)
plt.xticks(x_time[::5],x_tick_label)    #刻度标签替换时要保证个数对应，参1表示原刻度，参2表示刻度对应的标签
plt.yticks(y_tick)
#添加网格
plt.grid(linestyle='--',alpha=0.3)  #网格线段类型和透明度
#显示图像
plt.show()
```

```python
#添加标题，x轴、y轴标签，保存图片
#创建画布
plt.figure(figsize=(10,5),dpi=100)
#绘图
plt.plot(x_time,y_tem,color='red')
#添加刻度
x_tick_label=[f'11点{i}分' for i in range(0,60,5)]
y_tick=range(10,30,5)
plt.xticks(x_time[::5],x_tick_label)    #刻度标签替换时要保证个数对应，参1表示原刻度，参2表示刻度对应的标签
plt.yticks(y_tick)
#添加网格
plt.grid(linestyle='--',alpha=0.3)  #网格线段类型和透明度
#更改标题
plt.title('北京某天一小时内的温度变化折线图',fontsize=20)
#更改x，y轴标签
plt.xlabel('时间',fontsize=12)
plt.ylabel('温度',fontsize=12)
#保存图片
plt.savefig('./北京温度变化曲线.png')
#显示图像
plt.show()
```

```python
#在一个坐标轴中添加多个图像
#准备数据
x_time=range(0,24,1)
print(x_time)
y_bj=np.random.randint(20,30,24)
y_shh=np.random.uniform(10,20,24)
y_shh
#创建画布
plt.figure(figsize=(10,8),dpi=100)
#绘图
plt.plot(x_time,y_bj,color='red',label='北京')
plt.plot(x_time,y_shh,color='blue',label='上海',linestyle='--')
#创建x，y轴的刻度范围和刻度数据
x_tick_label=[f'{i}时00分' for i in range(0,24,2)]
y_tick=range(10,35,5)
plt.xticks(x_time[::2],x_tick_label)
plt.yticks(y_tick)
#添加标题和标签
plt.title('北京和上海24小时内的温度对比数据',fontsize=20)
plt.xlabel('时间(time)',fontsize=12)
plt.ylabel('温度(tem)',fontsize=12)
#添加图例
plt.legend(loc=0)
#添加网格
plt.grid(linestyle='-.',alpha=0.5)
#保存图片
plt.savefig('./北京和上海的温度对比.png')     #保存图片要写在显示图片之前
#显示图片
plt.show()
```

```python
#在一个画布中绘制多个图片
# axs1=plt.subplot(2,2,1)    #错误，subplot三个参数只能返回一个axes对象,subplot需要创建画布
# axs1.plot([1,2,3],[2,2,2])
fig,axs=plt.subplots(2,2,figsize=(20,16),dpi=100)       #subplots两个参数返回两个值，分别代表当前画布和绘图区域
axs[0,0].plot([1,2,3],[2,3,4])
axs[0,0].set_xlabel('x轴标签',fontsize=12)
axs[0,1].plot([1,2,3],[2,3,4])
axs[0,1].set_xticks(range(0,8,2),['A','B','C','D'])
axs[1,1].plot([1,2,3],[2,3,4])
axs[1,1].set_title('标题')
axs[1,0].plot([1,2,3],[2,3,4])
axs[1,0].grid()
plt.show()
```
