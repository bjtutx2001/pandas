## 1.1、创建ndarray对象


```python
#numpy是一个和数学操作相关的库，numpy的数组是多维的，并且是固定类型，数组关键字是ndarray，主要用于：矩阵运算
#创建ndarray对象，注意ndarray的下标以及从0开始
arr1 = np.arange(15).reshape(3, 5)
print(arr1)
```

## 1.2、演示numpy的常用属性

```python
print(f'numpy的轴:{arr1.ndim}')   #ndim 不是维度，是轴，也就是嵌套了几层
print(f'numpy的维度:{arr1.shape}')   #shape 维度，几行几列
print(f'numpy的元素类型:{arr1.dtype}')   #元素类型
print(f'numpy的元素个数:{arr1.size}')   #元素个数
print(f'numpy的元素占用字节数:{arr1.itemsize}')   #
```

## 1.3、常用的创建ndarray对象的几种方式

```python
#array()函数，把python列表转换为ndarray对象
my_list = [1,2,3,4,5,6,8]
arr2 = np.array(my_list)
print(arr2)
print(type(arr2))
```

```python
#arange()函数创建ndarray对象。格式arange(起始，结束，步长，dtype=类型)，类似于range函数，返回的是ndarray对象
arr3=np.arange(0,14,2,dtype=np.int32)
print(arr3)
print(type(arr3))
print(f'元素的类型：{arr3.dtype}')
```

```python
#随机数生成ndarray对象
#random.rand()生成0.0~1.0之间，包左不包右的小数矩阵
arr4=np.random.rand(3,5)    #3行5列
print(arr4)
print(type(arr4))
```

```python
#random.randint(起始值，结束值，size=（行，列）)生成指定范围，包左不包右的整数矩阵
arr5=np.random.randint(1,99,size=(5,5))    #3行5列
print(arr5)
print(type(arr5))
```

```python
#random.randuniform(起始值，结束值，size=（行，列）)生成指定范围，包左不包右的随机小数矩阵
arr6=np.random.uniform(1,99,size=(5,5))    #3行5列
print(arr6)
print(type(arr6))
```

## 1.5等差数列与等比数列

```python
#np.logspace(a,b,n,base=k)以a为开始，b为结束，取n个等间隔点，生成k为底的等比数列.base为设置底数，默认为10
a=np.logspace(0,9,10)
print(a)
b=np.logspace(0,9,10,base=2)
print(b)
```

```python
#np.linspace(a,b,n,endpoint=false)以a开始，b结束，取n个等间隔点，生成等差数列，endpoint决定了知否包含b
b=np.linspace(1,5,5,endpoint=True)  #包括尾数时，等比计算为(b-a)/(n-1)
print(b)
b=np.linspace(1,5,5,endpoint=False) #不包括尾数时，等比计算为(b-a)/n
print(b)
c=np.linspace(1,5,6,endpoint=False)
print(c)
```

## 1.6、numpy内置函数

### 常用函数

```python
#1、创建标准正态分布序列,如果想改变方差和标准差，就加常数乘系数
arr1=np.random.randn(3,5)
print(arr1)
```

```python
#2、测试基本函数
print(np.ceil(arr1))    #向上取整，比它大的最小的那个整数
print(np.floor(arr1))   #向下取整
print(np.rint(arr1))    #四舍五入
print(np.abs(arr1))     #绝对值
print(np.multiply(arr1,arr1)) #对应元素相乘
print(arr1*arr1)               #对应元素相乘
# print(np.matmul(arr1,arr1))     #矩阵乘法
# print(arr1@arr1)                #矩阵乘法
print(np.where(arr1>0,1,-1))     #如果元素大于0就变成1，否则变成负一，类似于三元表达式
```

### 统计函数

```python
arr2=np.arange(0,9).reshape(3,3)
print(arr2)
```

```python
print(np.cumsum(arr2))  #累加和，前面所有元素累加到当前元素
print(np.sum(arr2,axis=0))     #所有元素求和,axis=0表示按列求和
```

### 去重函数

```python
arr1 = np.array([[1,2,3],[3,3,5]])
print(arr1)
print(np.unique(arr1)) #去重，会返回新的数组
```

### 排序函数

```python
#sort函数，numpy的sort函数会返回一个新的数组，而python自带的sort是列表方法，会直接修改原列表
arr2=[1,54,2,3,4,5,19]
arr3=np.sort(arr2)
print(arr3)
arr2.sort()
print(arr2)
```
