# Pandas核心数据结构/类

### 核心数据结构：Series和Dataframe，用来组织和存放数据

# Series数据类型

```python
'''
Series可以理解为一行或一列数据，类似于一维数组，由索引index和值value组成
在Pandas中其实是没有行的概念的，任何数据取出来都是以列的形式
'''
import pandas as pd
import numpy as np
```

## 创建Series对象

```python
#自定义Series对象
s1=pd.Series([1,2,3,4,5])   #没有写index，就会默认自增索引
print(s1)
```

```python
#自定义Series对象，采用自定义索引
s2=pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])   #列表形式，常用形式
s22=pd.Series([i for i in range(5)],index=['a','b','c','d','e'])    #列表推导式
print(s22)
```

```python
#自定义Series对象，采用元组形式
s3=pd.Series((1,2,3,4,5),index=('a','b','c','d','e'))   #元组形式，元组是不可变对象，传给pandas之后，会被转换成index对象
print(s3)
#自定义Series对象，采用词典形式
s4=pd.Series({'a':1,'b':2,'c':3,'d':5})   #词典形式
print(s4)
```

```python
#采用numpy创建Series对象
s5=pd.Series(np.arange(5))
print(s5)
```

## Series对象的属性

```python
#构建Series对象
s6=pd.Series(data=[i for i in range(6)],index=[j for j in 'ABCDEF'])    #列表推导式
print(s6)
```

```python
#获取Series对象的索引列
print(s6.index)
```

```python
#获取Series对象的值
print(s6.values)
```

```python
#根据索引获取元素
print(s6['D'])
```

```python
#根据索引修改Series对象的元素值
s6['A']=39
print(s6)
```

# Dataframe数据类型

```python
#Dataframe是一个类似于二维数组或表格的对象，即有行索引（index），又有列索引（solumns）
#axis=0，代表行索引，sum(axis=0)代表，按上下的方向求和，最后得到一行，axis=1则代表按左右的方向进行操作，最后得到一列
```

## 创建Dataframe对象

```python
#采用字典+列表形式创建，在字典里面嵌套列表 （最推荐）
info = {
    '姓名':['zbw','xjx','myy','shy'],
    '年龄':[18,18,18,18],
    '性别':['男','女','男','女'],
}
#将上述数据封装成Dataframe对象
d1=pd.DataFrame(data=info,index=['a','b','c','d'])  #默认情况下，字典的键名就是列名，值就是列数据
print(d1)
```

```python
#字典套字典套列表
info={'a':{'姓名':'zbw','年龄':18,'性别':'男'},
      'b':{'姓名':'myy','年龄':18,'性别':'男'},
      'c':{'姓名':'xjx','年龄':18,'性别':'女'},        #字典套字典的情况下，外层字典的键名为列名。内层字典的键名为行名,一个内层字典就是一列数据
}
d2=pd.DataFrame(data=info)
print(d2)
```

```python
#列表套元组、列表套列表
info=[['zbw',18,'nan'],
      ['myy',18,'nan'],
      ['xjx',18,'nv'],
]
info=[  ('zbw',18,'nan'),
        ('myy',18,'nan'),
        ('xjx',18,'nv'),
]
d3=pd.DataFrame(data=info,index=['a','c','b'],columns=['姓名','年龄','性别'])
print(d3)
```

```python
#通过numpy的ndarray对象，创建Dataframe对象
#创建ndarray对象
a1=np.arange(0,12).reshape(3,4)
print(a1)
d4=pd.DataFrame(data=a1,index=[1,2,3],columns=['a','b','c','d'])
print(d4)
```

## Dataframe数据类型简单案例

```python
#生成10名同学5门功课的成绩，范围40-100
info=np.random.randint(40,101,(10,5))   #左闭右开
print(info)
```

```python
#封装为Dataframe对象
score_list=pd.DataFrame(info)
print(score_list)
```

```python
#修改索引和列名字
column_names=['语文','数学','英语','物理','化学']
index_names=['同学'+str(i) for i in range(1,score_list.shape[0]+1)]
#方法1
# score_list.index=index_names
# score_list.columns=column_names
#方法2,通过rename精准修改，需要传入字典
score_list.rename(columns={i: column_names[i] for i in range(len(column_names))}, index={i: index_names[i] for i in range(len(index_names))}, inplace=True)     #字典推导式 i：column_names[i] for i in 10 字典的键值对都会遍历
# print(score_list)
score_list
```

```python
score_list.shape[0]
```

## dataFrame基本属性

```python
#基本属性包括：纬度（行列数），索引列，列名，数据，行列转置等等
print('score_list的维度为：',score_list.shape)   #shape属性，返回一个元组，包含行数和列数
print('score_list的行数为：',score_list.shape[0])   #shape属性，返回一个元组，包含行数和列数
print('score_list的列数为：',score_list.shape[1])   #shape属性，返回一个元组，包含行数和列数
print('score_list的索引列为：',score_list.index)   #index属性，返回索引列
print('score_list的列名为：',score_list.columns)   #columns属性，返回列名
print('score_list的数据为：',score_list.values)       #data属性，返回数据
print('score_list的行列转置为：',score_list.T)       #T属性，返回行列转置
```

# Pandas的数据类型dtype

```python
#pandas的数据类型和python基本一样，只有几个不同，例如：object对应于python的str类型，np.nan/pd.NA对应于python的None类型，datetime64对应于python的datetime类型，timedelta[ns]对应于python的timedelta类型
#此外，pandas还支持一些特殊的数据类型，例如：category类型，用于表示分类数据，sparse类型，用于表示稀疏数据，interval类型，用于表示区间数据，period类型，用于表示时间段数据，等等
```

```python
#创建dataframe对象，演示缺省数据类型
df=pd.DataFrame(
    {'姓名':{'学生1':'zbw','学生2':'myy','学生3':'xjx'},
     '年龄':{'学生1':18,'学生2':18,'学生3':18},
     '性别':{'学生1':'男','学生2':'男','学生3':'女'}
    }
)   #字典套字典形式
df
```

```python
#演示时间差timedelta类型，时间差的常用定义格式为：xx days xx:xx:xx
df4=pd.to_datetime(['2025-2-18 00:30:56'])
df4 #默认只显示最后一个表达式，后面还有语句运行就不显示
print(df4.dtype)

df5=pd.to_datetime(['2026-10-10 15:38:36'])
df5
df6=df5-df4
print(df6)
print(df6.dtype)
```

# Pandas基本操作

## 加载数据

```python
#加载股票数据
stock_info=pd.read_csv('./stock_day.csv')
stock_info
```

```python
#移除不需要的列
df_st=stock_info.drop(columns=['ma5','ma10','ma20','v_ma5','v_ma10','v_ma20','turnover'],inplace=False)
df_st
df_st.info()    #详细信息
df_st.describe() #统计信息
```

## 索引操作

```python
#根据行列索引获取数据
df=df_st['open']
df.info()   #结果是series对象，因为dataframe的每一行/列都是series对象
print(df_st['open']['2018-02-13'])  #必须先列后行
# df_st['2018-02-13']['open']  #错误
```

```python
#通过loc和iloc获取数据，可进行切片,先行后列
#loc是通过标签名字获取数据
df_st.loc['2018-02-13','open']   #获取指定行数据
df_st.loc['2018-02-13':'2018-02-27':-1,'open':'volume':2]   #通过切片获取数据，步长为-1表示倒序，执行切片时，看的是标签在原表中的位置，步长为2表示每隔一行/列取一个数据，loc的切片是包含尾数的，iloc的切片是不包含尾数的
```

```python
#iloc是通过位置索引获取数据
df_st.iloc[0,0]   #获取指定位置数据 第0行第0列
df_st.iloc[0:5,0:3]   #通过切片获取数据，前5行，前3列，iloc的切片是不包含尾数的，同样可以倒序
```

## 赋值操作

```python
#查看源数据
# df_st1=df_st    #只是指向同一个对象，修改df_st1会影响df_st
df_st1=df_st.copy()  #创建副本，避免影响原数据
df_st1
#赋值
df_st1['open']=23
df_st1
df_st1.low=1    #用.的方式和中括号效果一致，但如果列名中有空格或特殊字符，就不能用.的方式赋值
df_st
```

```python
#按照值排序
df_st.sort_values('open',ascending=False)   #按照open列的值进行排序，ascending=False表示降序，默认是升序
df_st.sort_values(['open','high'],ascending=[False,True])   #先按照open列降序排序，如果open列相同，再按照high列升序排序
```

```python
#按照索引排序
df_st.sort_index(ascending=False)   #按照索引排序，ascending=False表示降
#series也可以按照索引排序
```

## 四则运算

```python
#加减乘除
df_st['open']+1
df_st['open'].add(2)
```

```python
#逻辑运算
#筛选出open大于25的值
df_st[df_st['open']>25]
```

```python
#筛选出close大于23小于25的值
#方法1：采用逻辑运算符&，需要加括号
df_st[(df_st['close']>23) & (df_st['close']<25)]
#方法2：采用query方法
df_st.query('close>23 and close<25')   #query方法中，query方法中可以直接使用列名，而不需要加引号
```

```python
#固定值筛选，选出high等于23.78或24.74的值
df_st[(df_st['high']==23.78) | (df_st['high']==24.74)]
df_st.query('high==23.78 or high==24.74')
df_st.query('high in [23.78,24.74]')   #in方法，判断是否在列表中
```
